```python
import numpy as np
import matplotlib.pyplot as plt

def gerschgorin_disks(A):
    """ Computes the Gerschgorin disks for a square matrix A. """
    n = A.shape[0]
    centers = np.diag(A)  # Diagonal elements
    radii = np.sum(np.abs(A), axis=1) - np.abs(centers)  # Row sum excluding diagonal
    return centers, radii

def verify_gerschgorin(A):
    """ Verifies if the eigenvalues of A lie within the Gerschgorin disks. """
    centers, radii = gerschgorin_disks(A)
    eigenvalues = np.linalg.eigvals(A)

    print("Eigenvalues of A:", eigenvalues)
    for i, eig in enumerate(eigenvalues):
        is_inside = any(abs(eig - centers[j]) <= radii[j] for j in range(len(centers)))
        print(f"Eigenvalue {eig}: {'Inside' if is_inside else 'Outside'} Gerschgorin Disks")

def plot_gerschgorin(A):
    """ Plots the Gerschgorin disks along with the eigenvalues of A. """
    centers, radii = gerschgorin_disks(A)
    eigenvalues = np.linalg.eigvals(A)

    fig, ax = plt.subplots()

    # Plot Gerschgorin circles
    for i in range(len(centers)):
        circle = plt.Circle(
            (centers[i].real, centers[i].imag),
            radii[i],
            color='b',
            fill=False,
            linestyle='dashed'
        )
        ax.add_patch(circle)
        plt.plot(
            centers[i].real,
            centers[i].imag,
            'bo',
            markersize=8
        )  # Center of the disk

    # Plot eigenvalues
    plt.scatter(
        eigenvalues.real,
        eigenvalues.imag,
        color='r',
        label="Eigenvalues",
        marker='x'
    )

    ax.set_xlabel('Real')
    ax.set_ylabel('Imaginary')
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_title('Gerschgorin Disks & Eigenvalues')
    ax.set_xlim(
        min(centers) - max(radii) - 1,
        max(centers) + max(radii) + 1
    )
    ax.set_ylim(-max(radii) - 1, max(radii) + 1)
    ax.set_aspect('equal')
    plt.legend()
    plt.grid(True)
    plt.show()

# Example usage
A = np.array([[0, -1, -2],
              [1, 1, -1],
              [2, 1, 2]], dtype=float)

print("Matrix A:\n", A)
verify_gerschgorin(A)
plot_gerschgorin(A)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def gerschgorin_disks(A):
    centers = np.diag(A)
    radii = np.sum(np.abs(A), axis=1) - np.abs(centers)
    return centers, radii


def verify_gerschgorin(A):
    centers, radii = gerschgorin_disks(A)
    eigenvalues = np.linalg.eigvals(A)

    print("\nGERSCHGORIN THEOREM")
    print("=" * 50)

    print("\nMatrix A:")
    print(A)

    print("\nGerschgorin Disks:")
    for i in range(len(centers)):
        print(f"Disk {i+1}: Center = {centers[i]:.2f}, Radius = {radii[i]:.2f}")

    print("\nEigenvalues:")
    for i, eig in enumerate(eigenvalues):
        print(f"λ{i+1} = {eig.real:.4f} {eig.imag:+.4f}i")

    print("\nVerification:")
    for i, eig in enumerate(eigenvalues):
        is_inside = any(
            abs(eig - centers[j]) <= radii[j]
            for j in range(len(centers))
        )

        if is_inside:
            print(f"Eigenvalue λ{i+1} = {eig.real:.4f} {eig.imag:+.4f}i : INSIDE")
        else:
            print(f"Eigenvalue λ{i+1} = {eig.real:.4f} {eig.imag:+.4f}i : OUTSIDE")

    print("=" * 50)


def plot_gerschgorin(A):
    centers, radii = gerschgorin_disks(A)
    eigenvalues = np.linalg.eigvals(A)

    fig, ax = plt.subplots(figsize=(8, 6))

    # Plot Gerschgorin circles
    for i in range(len(centers)):
        circle = plt.Circle(
            (centers[i].real, centers[i].imag),
            radii[i],
            fill=False,
            linestyle='dashed',
            linewidth=2
        )

        ax.add_patch(circle)

        ax.plot(
            centers[i].real,
            centers[i].imag,
            'bo',
            markersize=7
        )

    # Plot eigenvalues
    ax.scatter(
        eigenvalues.real,
        eigenvalues.imag,
        color='red',
        marker='x',
        s=80,
        label='Eigenvalues'
    )

    ax.set_xlabel('Real')
    ax.set_ylabel('Imaginary')

    ax.axhline(0, linewidth=0.5)
    ax.axvline(0, linewidth=0.5)

    ax.set_title('Gerschgorin Disks & Eigenvalues')

    max_radius = max(radii)

    ax.set_xlim(
        min(centers) - max_radius - 1,
        max(centers) + max_radius + 1
    )

    ax.set_ylim(
        -max_radius - 1,
        max_radius + 1
    )

    ax.set_aspect('equal')
    ax.legend()
    ax.grid(True)

    plt.show()


# Example usage
A = np.array([
    [0, -1, -2],
    [1,  1, -1],
    [2,  1,  2]
], dtype=float)

verify_gerschgorin(A)
plot_gerschgorin(A)
```
