```python
#format string fmt:- marker /line/colour
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([0,2,3,5,7,10])
ypoints= np.array([0,4,9,25,49,100])
plt.plot(xpoints,ypoints,'*:b')# or '*--b'
plt.show()
```

Marker size : short form ms

```python
plt.plot(xpoints,ypoints,marker='o',ms=20)#ms se marker ka size bara kar lete hai
plt.show()
```

Marker color:- markeredgecolor---mec

```python
plt.plot(xpoints,ypoints,marker='*', ms=10,mec='r')
plt.show()

```

```python
plt.plot(xpoints,ypoints,marker='o', ms=10,mec='r')
plt.show()
```

Marker face color:-mfc

```python
plt.plot(xpoints,ypoints,marker='*', ms=10,mfc='r')
plt.show()
```

matplotlib line:-line style or ls

```python
plt.plot(xpoints,ypoints,linestyle='dashed')
plt.show()
```

or

```python
plt.plot(xpoints,ypoints,ls=':')
plt.show()
```

line color change

```python
plt.plot(xpoints,ypoints,color='r')
plt.show()
```

line size-linewidth se define

```python
plt.plot(xpoints,ypoints,linewidth='10.0')
plt.show()
```

Multiples line in the same plot

```python
import matplotlib.pyplot as plt
import numpy as np
x1=np.array([0,3,5,7,9,10])
y1=np.array([0,9,25,49,81,100])
y2=np.array([0,6,10,14,18,20])
plt.plot(x1,y1,x1,y2)
plt.title('plot showing two function')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x1=np.array([0,3,5,7,9,10])
y1=np.array([0,9,25,49,81,100])
y2=np.array([0,6,10,14,18,20])
font1={'family':'serif','color':'black','size':20}
font2={'family':'serif','color':'pink','size':30}
plt.plot(x1,y1,x1,y2)
plt.title('plot showing two function', fontdict=font1)
plt.xlabel('X-axis',fontdict=font1)
plt.ylabel('Y-axis',fontdict=font2)
plt.show()
```

subplots: are shown with matrix

```python
import matplotlib.pyplot as plt
import numpy as np
#plot 1
x1=np.array([0,2,4,6,8])
y1=np.array([0,4,16,36,64])
#subplot
plt.subplot(2,1,1)#matrx define ke and uske 2 rows and 1 col
plt.plot(x1,y1)
plt.title('plot-1')
#data 2
x2=np.array([4,5,6,7,8])
y2=np.array([8,10,12,14,16])
#subplot
plt.subplot(2,1,2)
plt.plot(x2,y2)
plt.title('plot-2')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
#plot 1
x1=np.array([0,2,4,6,8])
y1=np.array([0,4,16,36,64])
#subplot
plt.subplot(1,2,1)#matrx define ke and uske 2 rows and 1 col
plt.plot(x1,y1)
plt.title('plot-1')
#data 2
x2=np.array([4,5,6,7,8])
y2=np.array([8,10,12,14,16])
#subplot
plt.subplot(1,2,2)
plt.plot(x2,y2)
plt.title('plot-2')
plt.show()
```


Scatter plot

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2,4,10,5,-3,-5,8,20])
y=np.array([10,-2,34,-20,5,2,0,6])
plt.scatter(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Scatter plot')
plt.show()
```
