global variable:

```python
x="beautiful"
def myfunc():
    print("python is" + x)

myfunc()
```

```python
x="beautiful"
def myfunc():
    x="fantastic" #x is a local variable
    print("python is" + x)

myfunc()
print("python is" +x)
```

```python
def myfunc():
    global x # x is a global variable
    x="good"

myfunc()
print("python is " +x)

 
```

```python
x="beautiful"
print(x)
def myfunc():
    global x
    x="fantastic" #x is a local variable
    print("python is " + x)

myfunc()
print("python is " +x)
```
